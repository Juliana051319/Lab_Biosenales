```python
import numpy as np
import matplotlib.pyplot as plt

# a) Creacion de los vectores
a = np.array([3.1, 1, -0.5, -3.2, 6])
b = np.array([1, 3, 2.2, 5.1, 1])
print(a,b)

# b) Producto escalar
prod_escalar = np.dot(a, b)
print("\nb) Producto escalar:", prod_escalar)

# c) Multiplicacion elemento a elemento
mult_elem = a * b
print("\nc) Multiplicacion punto a punto:", mult_elem)

# d) Matriz A
A = np.array([[2, -1, -3], [4, 1.5, -2.5], [7.3, -0.9, 0.2]])
print("\nd) Matriz A:\n", A)

# e) Transpuesta de A
A_T = A.T
print("\ne) Matriz transpuesta:\n", A_T)

# f) Ejemplos de funciones numpy
# ones: matriz de unos, round: redondea normal, ceil: redondea arriba, floor: redondea abajo
print("\nf) Funciones numpy:")
print("ones:\n", np.ones((2, 3)))
print("round:\n", np.round(A))
print("ceil:\n", np.ceil(A))
print("floor:\n", np.floor(A))

# g) Primera fila, tercera columna (indice 0 y 2)
print("\ng) Elemento fila 1, col 3:", A[0, 2])

# h) Segunda fila de la matriz A
print("\nh) Segunda fila:", A[1, :])
"\n"
# i) Dimensiones de la matriz A
# El atributo .shape retorna una tupla con (filas, columnas)
dimensiones = A.shape
print("i) Dimensiones de la matriz A (filas, columnas):", dimensiones)
"\n"
# j) Senoidal y[n] = sin(pi * 0.12 * n) en 0 <= n <= 100
# Definimos el vector de indices discretos n
n = np.arange(0, 101)  # Incluye el 100
y = np.sin(np.pi * 0.12 * n)
print("\nj) Primeros valores de y[n]:", y[:5])

"\n"
# k) Senoidal y2[n] = cos(2*pi * 0.03 * n) en el mismo intervalo
y2 = np.cos(2 * np.pi * 0.03 * n)
print("\nk) Primeros valores de y2[n]:", y2[:5])

# l) Creación de señales s[n] (suma) y t[n] (producto)
s = y + y2
t = y * y2

# m) Graficar y[n] y y2[n] en la misma figura (usando stem por ser señales discretas)
plt.figure(figsize=(10, 4))
plt.stem(
    n, y, linefmt="b-", markerfmt="bo", basefmt="r-", label="y[n] = sin(π*0.12*n)"
)
plt.stem(
    n,
    y2,
    linefmt="g-",
    markerfmt="go",
    basefmt="r-",
    label="y2[n] = cos(2π*0.03*n)",
)
plt.title("\nSeñales discretas y[n] y y2[n]")
plt.xlabel("Muestras (n)")
plt.ylabel("Amplitud")
plt.grid(True)
plt.legend()
plt.show()

# n) Graficar s[n] y t[n] en la misma figura
plt.figure(figsize=(10, 4))
plt.stem(n, s, linefmt="m-", markerfmt="mo", basefmt="r-", label="s[n] = Suma")
plt.stem(
    n, t, linefmt="c-", markerfmt="co", basefmt="r-", label="t[n] = Producto"
)
plt.title("\nOperaciones con señales: Suma s[n] y Producto t[n]")
plt.xlabel("Muestras (n)")
plt.ylabel("Amplitud")
plt.grid(True)
plt.legend()
plt.show()
```
